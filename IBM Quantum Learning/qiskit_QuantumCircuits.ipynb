### Qiskit 的电路量子比特排序约定

在 Qiskit 中，电路图中最顶部的量子比特具有索引 0，对应于量子比特元组中最右侧的位置（或对应于此元组的字符串、笛卡尔积或张量积中的位置）。倒数第二个量子比特具有索引 1, ，对应于元组中右数第二个位置，依此类推，一直到最底部的量子比特，该量子比特具有最高索引，对应于元组中最左侧的位置。

```python
# (q1,q0)
q1 = QuantumRegister(1, "q1")
q0 = QuantumRegister(1, "q0")
circuit = QuantumCircuit(q0,q1)
circuit.h(q0)
circuit.cx(q0, q1) # control bit, target bit

display(circuit.draw(output="mpl"))
display(Operator.from_circuit(circuit).draw("latex"))
```

```python
# 向线路输入 qubit：
zero = Statevector.from_label("0")
one = Statevector.from_label("1")
psi = zero.tensor(one)
display(psi.draw("latex"))

# 第一种写法
v = psi.evolve(circuit)
display(v.draw("latex"))
# 第二种写法
G = Operator.from_circuit(circuit)
v = psi.evolve(G)
display(v.draw("latex"))
```

```python
# (q2,q1,q0)
q2 = QuantumRegister(1, "q2")
q1 = QuantumRegister(1, "q1")
q0 = QuantumRegister(1, "q0")
circuit = QuantumCircuit(q0,q1,q2)
circuit.h(q0)
circuit.cx(q0, q2) # control bit, target bit
circuit.cx(q1, q2)

display(circuit.draw(output="mpl"))
display(Operator.from_circuit(circuit).draw("latex"))
```
